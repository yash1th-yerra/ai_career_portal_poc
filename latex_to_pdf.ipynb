````python
import shutil
import google.generativeai as genai
import json
import os
import tempfile
import subprocess

# --------------------------
# 🔑 Configure Gemini
# --------------------------
api_key = os.getenv("GEMINI_API_KEY")

if api_key:
    genai.configure(api_key=api_key)
    print("✅ Gemini API key configured successfully from environment variable.")
else:
    raise ValueError("❌ GEMINI_API_KEY environment variable not set.")


def ask_gemini(prompt, model="gemini-2.5-flash", temperature=0.3):
    """Call Gemini for resume optimization."""
    model = genai.GenerativeModel(model_name=model)
    response = model.generate_content(
        prompt,
        generation_config=genai.types.GenerationConfig(temperature=temperature)
    )
    return response.text


# --------------------------
# 🧠 Resume Optimizer
# --------------------------
def optimize_resume_with_gemini(job_description, base_latex, resume_json):
    """
    Optimizes a LaTeX resume using Gemini 2.5 based on:
    - Job description
    - Original LaTeX resume
    - Structured JSON (for factual grounding)
    """
    prompt = f"""
You are an expert technical resume optimizer and LaTeX editor.

You are given:
1. A candidate's LaTeX resume.
2. A job description.
3. A structured JSON version of the same resume.

Your task is to **refine the LaTeX resume** to make it more relevant and effective for the given job, while staying faithful to the JSON facts.

---

### STRICT INSTRUCTIONS

**1. STRUCTURE**
- Do NOT add or remove LaTeX sections or commands.
- Keep all \\section, \\item, and formatting unchanged.
- Only modify bullet points, ordering, and wording inside sections.
- Maintain reverse chronological order.

**2. WRITING STYLE**
- Start each bullet with a strong, non-repetitive action verb.
- Use concise, professional, results-focused phrasing.
- Avoid filler, buzzwords, or vague adjectives.
- Use active voice and past tense (unless current role).

**3. FACTUAL CONSISTENCY**
- Only use facts from the JSON.
- Do NOT invent new metrics, tools, or projects.

---

### JOB DESCRIPTION
{job_description}

### STRUCTURED RESUME JSON
{json.dumps(resume_json, indent=2)}

### ORIGINAL RESUME LATEX
{base_latex}

Now output only the optimized LaTeX resume (no markdown fences or commentary).
"""

    print("🤖 Optimizing resume with Gemini...")
    optimized = ask_gemini(prompt).strip()

    # Clean any accidental markdown fences
    optimized = optimized.replace("```latex", "").replace("```", "").strip()
    return optimized


# --------------------------
# 📄 LaTeX → PDF Compiler
# --------------------------
def latex_to_pdf(latex_code: str, output_pdf_path: str = "optimized_resume.pdf") -> str:
    """
    Compile LaTeX string to PDF using pdflatex (best for standard article templates).
    """
    import tempfile, subprocess, os, shutil

    with tempfile.TemporaryDirectory() as tmpdir:
        tex_path = os.path.join(tmpdir, "resume.tex")

        # Save LaTeX to temp file
        with open(tex_path, "w", encoding="utf-8") as f:
            f.write(latex_code)

        try:
            # Run pdflatex twice for references/links consistency
            for _ in range(2):
                result = subprocess.run(
                    ["pdflatex", "-interaction=nonstopmode", "resume.tex"],
                    cwd=tmpdir,
                    stdout=subprocess.PIPE,
                    stderr=subprocess.PIPE,
                    text=True
                )
                if result.returncode != 0:
                    raise subprocess.CalledProcessError(result.returncode, result.args, result.stdout, result.stderr)

            pdf_path = os.path.join(tmpdir, "resume.pdf")
            if os.path.exists(pdf_path):
                shutil.move(pdf_path, output_pdf_path)
                print(f"✅ PDF generated successfully at: {os.path.abspath(output_pdf_path)}")
                return os.path.abspath(output_pdf_path)
            else:
                print("⚠️ Compilation succeeded but PDF not found.")
                return None

        except subprocess.CalledProcessError as e:
            print("❌ LaTeX compilation failed!")
            print("---- STDOUT ----")
            print(e.stdout)
            print("---- STDERR ----")
            print(e.stderr)
            return None


# --------------------------
# 🚀 Main Flow
# --------------------------
if __name__ == "__main__":

    with open("resumes/Nagendra_structured.json", "r") as f:
        resume_json = json.load(f)




    optimized_resume = optimize_resume_with_gemini(job_description, latex_code, resume_json)

 
    pdf_path = latex_to_pdf(optimized_resume, "Nagendra_Optimized_Resume.pdf")

    if pdf_path:
        print(f"🎉 Final optimized resume PDF saved at: {pdf_path}")
    else:
        print("❌ Failed to generate optimized resume PDF.")
````
